## 2.3 Not luck (8 pts)

Pick one group difference from 2.2 (or a new one) and run the shuffle test from Day 3 on it. The function is written for you; it works on any grouping column and any numeric target.

Then two sentences in markdown: the gap you found, and what share of shuffles matched it. 0 to 1 percent means you found something. Over 5 percent means you haven't yet, and saying so is a full-credit answer.

```python
def shuffle_test(data, col, target, n=300):
    """How often does a SHUFFLED target produce a gap between groups as big as the real one?"""
    real = data.groupby(col, observed=True)[target].mean()
    gap = real.max() - real.min()
    shuffled = []
    for i in range(n):
        fake = data[target].sample(frac=1, random_state=i).values
        rates = data.assign(fake=fake).groupby(col, observed=True)["fake"].mean()
        shuffled.append(rates.max() - rates.min())
    share = np.mean([s >= gap for s in shuffled])
    print(f"{col}: our gap {gap:.3f}   shuffles at least as big: {share:.0%}")
    return share

# TODO: shuffle_test(df, "<your group column>", "<your target>")
```
