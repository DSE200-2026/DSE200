```python
# ▶ SETUP — downloads the three tables from Kaggle (no account needed)
import ast
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
import kagglehub

path = kagglehub.dataset_download("ihormuliar/starbucks-customer-data")
portfolio = pd.read_csv(f"{path}/portfolio.csv", index_col=0)
profile = pd.read_csv(f"{path}/profile.csv", index_col=0)
transcript = pd.read_csv(f"{path}/transcript.csv", index_col=0)
for name, d in [("portfolio", portfolio), ("profile", profile), ("transcript", transcript)]:
    print(f"{name:11} {d.shape}")

def shuffle_test(data, col, target, n=300):
    """How often does a SHUFFLED target produce a gap between groups as big as the real one?"""
    real = data.groupby(col, observed=True)[target].mean()
    gap = real.max() - real.min()
    shuffled = []
    for i in range(n):
        fake = data[target].sample(frac=1, random_state=i).values
        rates = data.assign(fake=fake).groupby(col, observed=True)["fake"].mean()
        shuffled.append(rates.max() - rates.min())
    share = np.mean([s >= gap for s in shuffled])
    print(f"{col}: our gap {gap:.3f}   shuffles at least as big: {share:.0%}")
    return share
```
