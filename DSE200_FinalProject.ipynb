```python
# ▶ SETUP — tools from class you'll want
import pandas as pd, numpy as np
import matplotlib.pyplot as plt

def check_join(left, right, on, how="left"):
    """Merge, and print the three-line check from Day 2 (4.14): rows before/after, unmatched, key uniqueness."""
    right_unique = not right.duplicated(on).any()
    out = left.merge(right, on=on, how=how, indicator=True)
    print(f"rows: left {len(left):,}  right {len(right):,}  ->  joined {len(out):,}")
    print("match:", out._merge.value_counts().to_dict())
    print("key unique on the right:", right_unique, "" if right_unique else "  <- rows will multiply")
    return out.drop(columns="_merge")

def shuffle_test(data, col, target, n=300):
    """How often does a SHUFFLED target produce a gap between groups as big as the real one?"""
    real = data.groupby(col, observed=True)[target].mean()
    gap = real.max() - real.min()
    shuffled = []
    for i in range(n):
        fake = data[target].sample(frac=1, random_state=i).values
        rates = data.assign(fake=fake).groupby(col, observed=True)["fake"].mean()
        shuffled.append(rates.max() - rates.min())
    share = np.mean([s >= gap for s in shuffled])
    print(f"{col}: our gap {gap:.3f}   shuffles at least as big: {share:.0%}")
    return share

print("check_join() and shuffle_test() ready.")
```
